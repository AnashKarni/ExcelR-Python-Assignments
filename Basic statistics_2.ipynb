```python
import pandas as pd
```

```python
# Load dataset

df = pd.read_csv("Patient_Data.csv")
billing = pd.read_csv("Billing_Data.csv")
```

# Task 1: Load and show summary using info()

```python
# info() shows column names, data types, and missing value counts
print ("Task 1: Dataset Info")
df.info()
```

# Task 2: Select only billing-related columns

```python
print("Task 2: Billing Columns Only")
billing_cols = df[['PatientID', 'Department', 'Doctor', 'BillAmount']]
print(billing_cols)
```

# Task 3: Drop unwanted admin columns

```python
print("Task 3: Drop Admin Columns")
df = df.drop(columns=['ReceptionistID', 'CheckInTime'])
print(df)
```

# Task 4: Total bill amount per department

```python
print("Task 4: Total bill amount per department")
result = df.groupby('Department')['BillAmount'].sum()
print(result)
```

# Task 5: Remove duplicate patient records

```python
print("Task 5: Remove Duplicates")
df = df.drop_duplicates(subset='PatientID').copy()
print(df)
```

# Task 6: Fill missing BillAmount with mean

```python
print("Task 6: Fill Missing BillAmount")
# Calculate average bill
mean_val = df['BillAmount'].mean()  

# Replace NaN with average
df['BillAmount'] = df['BillAmount'].fillna(mean_val)

print(df[['PatientID', 'BillAmount']])
```

# Task 7: Merge patient data with billing data

```python
print("Task 7: Merged Dataset")
merged = pd.merge(df, billing, on='PatientID')
print(merged)
```

# Task 8: Add new patients — Row-wise

```python
print("Task 8: Add New Patients(Row-wise)")
new_patients = pd.DataFrame({
    'PatientID': [106, 107],
    'Name':      ['Frank', 'Grace'],
    'Department':['Neurology', 'Cardiology'],
    'Doctor':    ['Dr. John', 'Dr. Smith'],
    'BillAmount':[4500, 3800]
})
df = pd.concat([df, new_patients], ignore_index=True)
print(df)
```

# Task 9: Add billing columns — Column-wise concat

```python
print("Task 9: Add Billing Columns (Column-wise)")
extra_cols = pd.DataFrame({
    'InsuranceCovered': [2000, 1500, 2500, 3000, 1000, 1800, 2200],
    'FinalAmount':      [3000, 3500, 5000, 3200, 4000, 2700, 1600]
})
# attach columns to the right
final_df = pd.concat([df, extra_cols], axis=1)  
print(final_df)
```

## Final Conclusion & Business Impact
1. **Data Quality Achieved:** Successfully transformed raw, fragmented hospital data into a clean, unified dataset by removing duplicates and imputing missing `BillAmount` records.
2. **Revenue Insights:** The `groupby` analysis reveals that the **Cardiology** department is currently driving the highest revenue initially, whereas missing records in other departments have now been standardized for accurate financial reporting.
3. **Readiness for Predictive Modeling:** With comprehensive patient history, insurance coverage, and final billing amounts merged into a single view, this dataset is now fully optimized to build dashboards for tracking doctor performance and department-wise profitability.
